### Viscous time scale of the disk

```python
au  = 1.49598e13     # Astronomical Unit       [cm]
GG  = 6.67408e-08    # Gravitational constant  [cm^3/g/s^2]
ms  = 1.98892e33     # Solar mass              [g]
mstar    = 1.08*ms #ms


r = 57*au
Omega_Kepler = np.sqrt(GG*mstar/r**3)
H = 0.069*r
alpha = 1e-2

nu = alpha*H**2*Omega_Kepler
### don't use r as local length scale; rather use the pressure scale height.
r_hill = r*(0.0044)**(1/3)

tau_visc = (4*r_hill)**2/nu

print(f"At alpha = {alpha}, r = {r/au}, we have a viscous time scale of {tau_visc} s")
print(f"That is equivalent to {tau_visc/60/60/24/365/1e6} Myrs.")
print(f"That is the equivalent of {tau_visc/(2*np.pi/Omega_Kepler)} orbits.")
```

## Viscous time scale from eq. 3 from Kanagawa 2017 

```python
au  = 1.49598e13     # Astronomical Unit       [cm]
GG  = 6.67408e-08    # Gravitational constant  [cm^3/g/s^2]
ms  = 1.98892e33     # Solar mass              [g]
mstar    = 1.08*ms #ms


r = 57*au
Omega_Kepler = np.sqrt(GG*mstar/r**3)
H = 0.069*r
alpha = 1e-2

Delta_gap = r *1.05

tau_visc = (Delta_gap/2/r)**2 *(H/r)**(-2)/alpha/Omega_Kepler

print(f"At alpha = {alpha}, r = {r/au}, we have a viscous time scale of {tau_visc} s")
print(f"That is equivalent to {tau_visc/60/60/24/365/1e6} Myrs.")
print(f"That is the equivalent of {tau_visc/(2*np.pi/Omega_Kepler)} orbits.")
```
